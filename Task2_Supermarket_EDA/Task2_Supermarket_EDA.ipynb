# Task 2: Exploratory Data Analysis

## Dataset: Supermarket Sales

### Objective
To perform Exploratory Data Analysis (EDA) on supermarket sales data using Python, Pandas, SQLite, and Matplotlib.

### Purpose
The analysis aims to understand sales patterns, customer behavior, product performance, payment methods, relationships between variables, and unusual patterns in the dataset.

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("SuperMarket Analysis.csv")
print("Dataset loaded successfully!")
```

```python
connection = sqlite3.connect("supermarket.db")

data.to_sql(
    "sales",
    connection,
    if_exists="replace",
    index=False
)

print("Supermarket database created successfully!")
```

```python
sales = pd.read_sql_query(
    "SELECT * FROM sales",
    connection
)

print("Sales data loaded from database!")
```

```python
sales.head()
```

```python
missing_values = sales.isnull().sum()

missing_values[missing_values > 0]
```

```python
sales.columns.tolist()
```

```python
sales["Date"] = pd.to_datetime(
    sales["Date"],
    errors="coerce"
)
```

## BUSINESS QUESTION 1
## Which branch has the highest sales?

```python
branch_sales = sales.groupby("Branch")["Sales"].sum().sort_values(
    ascending=False
)

branch_sales
```

## BUSINESS QUESTION 2
## Which product line generates the highest sales?

```python
product_sales = sales.groupby("Product line")["Sales"].sum().sort_values(
    ascending=False
)

product_sales
```

## BUSINESS QUESTION 3
## Which product line has the highest quantity sold?

```python
product_quantity = sales.groupby("Product line")["Quantity"].sum().sort_values(
    ascending=False
)

product_quantity
```

## BUSINESS QUESTION 4
## Which payment method is most popular?

```python
payment_counts = sales["Payment"].value_counts()

payment_counts
```

## BUSINESS QUESTION 5
## Which customer type is more common?

```python
customer_counts = sales["Customer type"].value_counts()

customer_counts
```

## BUSINESS QUESTION 6
## Compare sales by gender

```python
gender_sales = sales.groupby("Gender")["Sales"].sum()

gender_sales
```

## BUSINESS QUESTION 7
## Which branch has the highest average rating?

```python
branch_rating = sales.groupby("Branch")["Rating"].mean().sort_values(
    ascending=False
)

branch_rating
```

## Top 10 highest sales

```python
sales.nlargest(10, "Sales")
```

## Bottom 10 sales

```python
sales.nsmallest(10, "Sales")
```

## SALES TREND OVER TIME

```python
daily_sales = sales.groupby("Date")["Sales"].sum()
```

```python
daily_sales
```

## CHART 1
### Daily Sales Trend

```python
daily_sales.plot()

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

## CHART 2
## Sales by Branch

```python
branch_sales.plot(kind="bar")

plt.title("Total Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.show()
```

## CHART 3
## Sales by Product Line

```python
product_sales.plot(kind="bar")

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

## CHART 4
## Payment Methods

```python
payment_counts.plot(kind="bar")

plt.title("Payment Methods Used")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")
plt.show()
```

## CHART 5
## Quantity vs Total Sales

```python
plt.scatter(
    sales["Quantity"],
    sales["Sales"]
)

plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.show()
```

## CHART 6
## Distribution of Total Sales

```python
sales["Sales"].plot(
    kind="hist",
    bins=20
)

plt.title("Distribution of Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.show()
```

## OUTLIER ANALYSIS



```python
plt.boxplot(sales["Sales"].dropna())

plt.title("Total Sales Outliers")
plt.ylabel("Total Sales")
plt.show()
```

## CORRELATION

```python
sales.corr(numeric_only=True)
```

## SAVING THE CLEANED DATASET
