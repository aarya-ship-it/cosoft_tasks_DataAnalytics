## 7. Data Type Correction

```python
orders.dtypes
```

```python
date_columns=["OrderDate","RequiredDate","ShippedDate"] 
```

```python
for column in date_columns: orders[column]=pd.to_datetime(orders[column], errors="coerce")
```

```python
orders.dtypes
```

```python
orders["ShipCountry"].unique()
```

```python
orders["ShipCity"].unique()
```

```python
text_columns=orders.select_dtypes(include=["str"]).columns
```

```python
for columns in text_columns: orders[column]=orders[column]=orders[column].apply(lambda x:x.strip() if isinstance(x,str)else x)
```

```python
orders.isnull().sum()
```

```python
if orders["Freight"].isnull().sum() > 0:
    orders["Freight"] = orders["Freight"].fillna(
        orders["Freight"].median()
    )
```

```python
print("Dataset shape:")
print(orders.shape)

print("\nMissing values:")
print(orders.isnull().sum())

print("\nDuplicate records:")
print(orders.duplicated().sum())

print("\nData types:")
print(orders.dtypes)
```

## 8. Exporting the cleaned datset

```python
orders.to_csv(
    "orders_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
